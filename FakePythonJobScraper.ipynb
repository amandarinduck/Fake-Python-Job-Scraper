```python
import requests
from bs4 import BeautifulSoup
import re
#get html page w requests
url = requests.get('https://realpython.github.io/fake-jobs/')
soup = BeautifulSoup(url.text,'html.parser')

#pull relevant data
jobs = [x.text.strip() for x in soup.find_all('h2',class_="title")]
companies = [x.text.strip() for x in soup.find_all('h3',class_="subtitle")]
location = [x.text.strip() for x in soup.find_all('p',class_='location')]
links = [a['href'] for a in soup.find_all('a', href=re.compile(r"^https://realpython.github.io"))]
```

```python
#iterate through collected data and pass to a dataframe
import pandas as pd


data =[]
for i in range(len(jobs)):
    row = {
        "Position":jobs[i],
        "Company":companies[i],
        "Location":location[i],
        "Link":links[i]
    }
    data.append(row)

df = pd.DataFrame(data)

df.head()
```

```python
#export to csv
df.to_csv('jobs.csv')
```
